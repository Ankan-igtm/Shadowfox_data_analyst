```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/Raw Data.csv")

print("Shape:", df.shape)
display(df.head())
```

```python
categorical_columns = [
    "Attrition",
    "BusinessTravel",
    "Department",
    "EducationField",
    "Gender",
    "JobRole",
    "MaritalStatus",
    "OverTime"
]

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns

for col in numerical_columns:
    print(
        f"{col}: Min={df[col].min()}, Max={df[col].max()}"
    )
```

```python
for col in ["EmployeeCount", "Over18", "StandardHours"]:
    print(col, df[col].unique())
```

```python
df.drop(
    columns=["EmployeeCount", "Over18", "StandardHours"],
    inplace=True
)
```

```python
print(df.shape)
```

```python
print("Unique Employee Numbers:", df["EmployeeNumber"].nunique())
print("Total Employees:", len(df))
```

```python
print(df["Attrition"].value_counts())
```

```python
print(
    df["Attrition"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=[
        "Under 25",
        "25-34",
        "35-44",
        "45-54",
        "55+"
    ]
)
```

```python
df["Salary Group"] = pd.cut(
    df["MonthlyIncome"],
    bins=[0, 3000, 6000, 10000, float("inf")],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
df["Tenure Group"] = pd.cut(
    df["YearsAtCompany"],
    bins=[-1, 2, 5, 10, 20, float("inf")],
    labels=[
        "0-2 Years",
        "3-5 Years",
        "6-10 Years",
        "11-20 Years",
        "20+ Years"
    ]
)
```

```python
df["Attrition Flag"] = df["Attrition"].map({
    "Yes": 1,
    "No": 0
})
```

```python
df["Attrition Flag"].value_counts()
```

```python
df["OverTime Flag"] = df["OverTime"].map({
    "Yes": 1,
    "No": 0
})
```

```python
df["Company Tenure %"] = np.where(
    df["TotalWorkingYears"] > 0,
    (df["YearsAtCompany"] / df["TotalWorkingYears"]) * 100,
    0
)
```

```python
print("Final Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nColumns:")
print(df.columns.tolist())
```

```python
output_file = "IBM_HR_Analytics_Cleaned.csv"

df.to_csv(output_file, index=False)

print(f"Saved: {output_file}")
```
