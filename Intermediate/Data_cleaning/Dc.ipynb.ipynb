```python
import pandas as pd
```

```python
file_path = "/content/SuperStore_Sales_Dataset_Raw.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Dataset Information:\n")
df.info()

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python
df.columns = df.columns.str.strip()

df.rename(
    columns={
        "Row ID+O6G3A1:R6": "Row ID"
    },
    inplace=True
)

print(df.columns.tolist())
```

```python
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values[missing_values > 0])
```

```python
df.drop(
    columns=["ind1", "ind2"],
    inplace=True
)
```

```python
print(df.columns.tolist())
print("New shape:", df.shape)
```

```python
df["Return Status"] = df["Returns"].apply(
    lambda x: "Returned" if x == 1 else "Not Returned"
)
```

```python
print(df["Return Status"].value_counts())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)

print("Shape after removing duplicates:", df.shape)
```

```python
print("Unique Order IDs:", df["Order ID"].nunique())
print("Total Rows:", len(df))
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    format="%d-%m-%Y",
    errors="coerce"
)
```

```python
print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
print(df[["Order Date", "Ship Date"]].isnull().sum())
```

```python
invalid_dates = df[
    df["Ship Date"] < df["Order Date"]
]

print("Invalid Shipping Date Records:")
print(len(invalid_dates))
```

```python
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

```python
df[[
    "Order Date",
    "Ship Date",
    "Shipping Days"
]].head()
```

```python
numeric_columns = [
    "Sales",
    "Quantity",
    "Profit",
    "Shipping Days"
]

print(df[numeric_columns].describe())
```

```python
invalid_sales = df[df["Sales"] <= 0]

print("Invalid Sales Records:", len(invalid_sales))
display(invalid_sales.head())
```

```python
negative_profit = df[df["Profit"] < 0]

print("Negative Profit Records:", len(negative_profit))

display(
    negative_profit[
        [
            "Order ID",
            "Product Name",
            "Category",
            "Sales",
            "Profit"
        ]
    ].head()
)
```

```python
print("Sales Outliers:")
print(df["Sales"].describe())

print("\nProfit Outliers:")
print(df["Profit"].describe())

print("\nQuantity:")
print(df["Quantity"].describe())
```

```python
categorical_columns = [
    "Ship Mode",
    "Segment",
    "Country",
    "Region",
    "Category",
    "Sub-Category",
    "Payment Mode",
    "Return Status"
]

for column in categorical_columns:
    print("\n" + "=" * 50)
    print(column)
    print(df[column].value_counts())
```

```python
text_columns = df.select_dtypes(
    include="object"
).columns

for column in text_columns:
    df[column] = df[column].astype(str).str.strip()
```

```python
text_columns = df.select_dtypes(
    include="object"
).columns

for column in text_columns:
    df[column] = df[column].astype(str).str.strip()
```

```python
df["Year"] = df["Order Date"].dt.year

df["Month Number"] = df["Order Date"].dt.month

df["Month"] = df["Order Date"].dt.month_name()

df["Month Year"] = df["Order Date"].dt.to_period("M").astype(str)
```

```python
df[
    [
        "Order Date",
        "Year",
        "Month Number",
        "Month",
        "Month Year"
    ]
].head()
```

```python
df["Profit Margin %"] = (
    df["Profit"] / df["Sales"] * 100
)
```

```python
df["Profit Margin %"] = df[
    "Profit Margin %"
].round(2)
```

```python
df[
    [
        "Sales",
        "Profit",
        "Profit Margin %"
    ]
].head()
```

```python
print("Final Missing Values:")
print(
    df.isnull().sum()[
        df.isnull().sum() > 0
    ]
)
```

```python
import os

output_file = "/mnt/data/SuperStore_Sales_Dataset_Cleaned.csv"

# Extract the directory path
output_dir = os.path.dirname(output_file)

# Create the directory if it doesn't exist
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
    print(f"Created directory: {output_dir}")

df.to_csv(
    output_file,
    index=False
)

print("Cleaned dataset saved successfully!")
print(output_file)
```
